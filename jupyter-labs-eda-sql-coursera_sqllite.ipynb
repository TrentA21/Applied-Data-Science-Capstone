## Tasks

Now write and execute SQL queries to solve the assignment tasks.

### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
%sql SELECT DISTINCT Launch_Site FROM SPACEXTABLE;
```

> *Explanation (Task 1):* `DISTINCT` returns each launch-site name once → **4 sites: CCAFS LC-40, VAFB SLC-4E, KSC LC-39A, CCAFS SLC-40** (CCAFS LC-40 and CCAFS SLC-40 are the same pad under its old and new name).


### Task 2


#####  Display 5 records where launch sites begin with the string 'CCA' 


```python
%sql SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5;
```

> *Explanation (Task 2):* `LIKE 'CCA%'` matches any site starting with *CCA*; `LIMIT 5` keeps the first five rows (the earliest Cape Canaveral launches, 2010–2013).

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
%sql SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_Mass_kg FROM SPACEXTABLE WHERE Customer = 'NASA (CRS)';
```

> *Explanation (Task 3):* `SUM` adds the payload mass of every launch whose customer is exactly `NASA (CRS)` → **45,596 kg**.

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
%sql SELECT AVG(PAYLOAD_MASS__KG_) AS Avg_Payload_Mass_kg FROM SPACEXTABLE WHERE Booster_Version = 'F9 v1.1';
```

> *Explanation (Task 4):* `AVG` over rows whose booster is exactly `F9 v1.1` → **2,928.4 kg**.

### Task 5

##### List the date when the first successful landing outcome in ground pad was acheived.


_Hint:Use min function_ 


```python
%sql SELECT MIN(Date) AS First_Successful_Ground_Pad_Landing FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (ground pad)';
```

> *Explanation (Task 5):* `MIN(Date)` among successful ground-pad landings → **2015-12-22** (Orbcomm OG2 mission, the first Falcon 9 landing on land).

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
%%sql
SELECT Booster_Version FROM SPACEXTABLE
WHERE Landing_Outcome = 'Success (drone ship)'
  AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000;
```

> *Explanation (Task 6):* Combines three conditions with `AND`: drone-ship success and payload strictly between 4,000 and 6,000 kg → **F9 FT B1022, F9 FT B1026, F9 FT B1021.2, F9 FT B1031.2**.

### Task 7




##### List the total number of successful and failure mission outcomes


```python
%%sql
SELECT Mission_Outcome, COUNT(*) AS Total FROM SPACEXTABLE GROUP BY Mission_Outcome;
```

> *Explanation (Task 7):* `GROUP BY Mission_Outcome` with `COUNT(*)` → **100 successful** missions (98 `Success` + 1 `Success` with a trailing space + 1 `Success (payload status unclear)`) and **1 failure** (`Failure (in flight)`).

### Task 8



##### List the   names of the booster_versions which have carried the maximum payload mass. Use a subquery


```python
%%sql
SELECT DISTINCT Booster_Version FROM SPACEXTABLE
WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE);
```

> *Explanation (Task 8):* The subquery finds the largest payload (15,600 kg); the outer query lists every booster that carried it → **12 Falcon 9 Block 5 boosters** (B1048.4, B1049.4, B1051.3, B1056.4, B1048.5, B1051.4, B1049.5, B1060.2, B1058.3, B1051.6, B1060.3, B1049.7) — all Starlink missions.

### Task 9


##### List the failed landing_outcomes in drone ship, their booster versions, and launch site names for in year 2015

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6, 2) as month to get the months and substr(Date,1,4)='2015' for year.**


```python
%%sql
SELECT substr(Date, 6, 2) AS Month, Landing_Outcome, Booster_Version, Launch_Site
FROM SPACEXTABLE
WHERE Landing_Outcome = 'Failure (drone ship)' AND substr(Date, 1, 4) = '2015';
```

> *Explanation (Task 9):* SQLite has no month names, so `substr(Date,6,2)` gives the month and `substr(Date,1,4)='2015'` filters the year → **2 failed drone-ship landings in 2015**: January (F9 v1.1 B1012) and April (F9 v1.1 B1015), both from CCAFS LC-40.

### Task 10

##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order
**Note: SQLLite does not support monthnames. So you need to use substr(Date,9,2) as date to get the date,  substr(Date, 6, 2) as month to get the months and substr(Date,1,4)='2015' for year.You need to use string format for date as '20100604' and '20170320'**


```python
%%sql
SELECT Landing_Outcome, COUNT(*) AS Outcome_Count
FROM SPACEXTABLE
WHERE Date BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY Landing_Outcome
ORDER BY Outcome_Count DESC;
```

> *Explanation (Task 10):* Filters to 2010-06-04 … 2017-03-20, counts each landing outcome and sorts high→low. **No attempt (10)** is the most common, then Success (drone ship) 5, Failure (drone ship) 5, Success (ground pad) 3, Controlled (ocean) 3, Uncontrolled (ocean) 2, Failure (parachute) 2, Precluded (drone ship) 1.
